```python
# Pareto Analysis (80/20 Rule)

pareto = df.groupby('product_detail')['Revenue'].sum().reset_index()

pareto = pareto.sort_values(by='Revenue', ascending=False)

pareto['Cumulative_Revenue'] = pareto['Revenue'].cumsum()

pareto['Cumulative_Percentage'] = (
    pareto['Cumulative_Revenue'] / pareto['Revenue'].sum()
) * 100

pareto.head(20)
```

```python
# Products Contributing to 80% Revenue

pareto[pareto['Cumulative_Percentage'] <= 80]
```

```python
len(pareto[pareto['Cumulative_Percentage'] <= 80])
```
